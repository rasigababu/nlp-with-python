```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report
```

```python
url = "https://raw.githubusercontent.com/justmarkham/pycon-2016-tutorial/master/data/sms.tsv"

df = pd.read_csv(url, sep='\t', names=['Category','Message'])


df.head()
```

```python
df.shape
```

```python
df['Category'].value_counts()
```

```python
df['Spam']=df['Category'].map({"ham":0,"spam":1})
```

```python
df['Spam'].head(10)
```

```python
df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(df['Message'],df['Spam'],test_size=0.2,random_state=42)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
model = Pipeline([
    ('vectorizer', CountVectorizer()),
    ('nb', MultinomialNB())
])

model.fit(X_train, y_train)

# Predict
y_pred = model.predict(X_test)

# Evaluate
print(classification_report(y_test, y_pred))
```

```python
messages = [
    "Hey, are you coming today?",
    "Win free recharge now!!!",
    "Let's meet at 5 PM"
]

pred = model.predict(messages)

for msg, p in zip(messages, pred):
    print(msg, "→", "Spam" if p==1 else "Ham")
```

```python
import joblib

# Save model
joblib.dump(model, "spam_model.pkl")
```

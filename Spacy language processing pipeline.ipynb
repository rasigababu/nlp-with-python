```python
import spacy
```

```python
nlp = spacy.blank('en')
```

```python
doc=nlp("Captain America ate 100$ of samosas. He then said he could do it all day.")
for token in doc:
    print(token)
```

```python
nlp.pipe_names
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
nlp.pipe_names
```

```python
doc=nlp("Captain America ate 100$ of samosas. He then said he could do it all day.")
for token in doc:
    print(token," | ",token.pos_,"| ",token.lemma_)
```

```python
doc = nlp("Tesla is going to acquire twitter for $45 billion")
for ent in doc.ents:
    print(ent.text,ent.label_,spacy.explain(ent.label_))
```

```python
from spacy import displacy

displacy.render(doc,style="ent")
```

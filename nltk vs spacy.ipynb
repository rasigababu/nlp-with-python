```python
spacy.load("en_core_web_sm")
```

```python
nlp = spacy.load("en_core_web_sm")
doc = nlp("Dr. Strange loves pav bhaji of mumbai. Hulk loves chat of Delhi.")
```

```python
# sentence tokenization
for sentences in doc.sents:
    print(sentences)
```

```python
#word tokenization
for sentences in doc.sents:
    for word in sentences:
      print(word)
```

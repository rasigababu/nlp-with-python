```python
from spacy.symbols import ORTH

nlp.tokenizer.add_special_case(
    "gimme",[
        {ORTH:"gim"},
        {ORTH: "me"}
    ]
)
text = "gimme me double cheese extra helathy pizza"
doc = nlp(text)
```

```python
for token in doc:
    print(token.text)
```

```python
import spacy
nlp=spacy.load("en_core_web_sm")
```

```python
nlp.pipe_names
```

```python
doc=nlp("Tesla is going to acquire Twitter.inc for $45 billion on 1 Feb 7am")
```

```python
for ent in doc.ents:
    print(ent.text,ent.label_,spacy.explain(ent.label_))
```

```python
from spacy import displacy
displacy.render(doc,style="ent")
```
